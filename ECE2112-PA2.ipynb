```python
import numpy as np
```

# Problem 1

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
X_normalized = (X-np.mean(X))/np.std(X)
X_normalized
```

```python
# all results compiled

print(X)
print(X.mean())
print(X.std())
print(X_normalized)
print(X_normalized.mean())
print(X_normalized.std())
```

```python
np.save("X_normalized.npy", X_normalized)
```

# Problem 2

```python
Y = np.arange(1,101)
Y = np.power(Y, 3)
Y.resize(10,10)
```

```python
C = Y
C
```

```python
div_by_4 = C[C%4==0]
```

```python
# all results compiled
print(C.shape)
print(div_by_4)
print(div_by_4.size)
```

```python
np.save("div_by_4.npy", div_by_4)
```

# Problem 3

```python
S = np.arange(1,37)
S = np.power(S, 2)
S.resize(6,6)
S
```

```python
S_mean = S.mean()
S_mean
```

```python
above_mean = S[S>S_mean]
```

```python
# all results compiled

print(S)
print(S_mean)
print(above_mean)
print(above_mean.size)
```

```python
np.save("above_mean.npy", above_mean)
```
